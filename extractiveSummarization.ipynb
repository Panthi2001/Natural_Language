```python
import nltk
import string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize,sent_tokenize

# Download resources if needed
nltk.download('punkt')
nltk.download('stopwords')
from heapq import nlargest
```

```python
text="""Over the past two decades, social media has revolutionized the way people communicate. Platforms such as Facebook, Twitter, Instagram, and TikTok have created virtual spaces where individuals from different corners of the world can connect, share ideas, and build communities. The emergence of these platforms has significantly altered the landscape of interpersonal communication, both positively and negatively.

One of the major benefits of social media is its ability to facilitate instant communication. Long-distance relationships, whether romantic, familial, or professional, can now be sustained with relative ease. Video calls, messaging apps, and status updates allow users to remain connected and informed in real-time. This has been particularly helpful during global crises, such as the COVID-19 pandemic, when physical distancing made traditional face-to-face communication difficult.

Social media also provides a platform for marginalized voices and social movements. Hashtags and viral posts have played critical roles in raising awareness about injustices and prompting collective action. Movements like #BlackLivesMatter, #MeToo, and climate change advocacy have gained traction globally, largely due to the widespread reach of social media.

However, the influence of social media is not entirely positive. The rise of online communication has led to a decline in face-to-face interactions, which are essential for developing empathy and nuanced social skills. People may become more isolated or anxious despite being constantly "connected." Moreover, the addictive design of many platforms can contribute to decreased attention spans and a constant need for external validation through likes, comments, and shares.

Another concerning aspect is the spread of misinformation. Algorithms that favor engagement often amplify sensational or false content, leading to echo chambers where users are exposed only to viewpoints similar to their own. This can fuel polarization and hinder constructive dialogue.

In conclusion, while social media has democratized communication and empowered many, it also presents challenges that society must navigate carefully. Striking a balance between digital interaction and real-world connection, and ensuring the ethical use of these platforms, will be key to fostering healthy communication in the future.

"""
```

```python
# Tokenize
words = word_tokenize(text)

# Define stopwords and punctuation
stop_words = set(stopwords.words('english'))
punctuations = set(string.punctuation)

#Word frequency
word_frequency={}

for word in words:
    lower_word = word.lower()
    if lower_word not in stop_words and word not in punctuations:
        word_count=word_frequency.get(word)
        if word_count is None:
            word_frequency[word]=1
        else:
            word_frequency[word]=word_count+1
        

print(word_frequency)
most_frequent = max(word_frequency, key=word_frequency.get)
max_frequency=word_frequency[most_frequent]
print(f"Most frequent word: '{most_frequent}' with frequency {max_frequency}")
```

```python
#Normalize frequency
for word,frequency in word_frequency.items():
    word_frequency[word]= frequency/max_frequency

print(word_frequency)
```

```python
#Iterate through sentences to get highest scored sentences
sentences=sent_tokenize(text)

scored_sentences={}

for sentence in sentences:
    words = word_tokenize(sentence)
    sentence_score=0
    for word in words:
        word_score=word_frequency.get(word)
        if word_score is None:
            sentence_score+=0
        else:
            sentence_score+=word_score
    scored_sentences[sentence]=sentence_score


print(scored_sentences)
```

```python


summarized_length=int(len(sentences)*0.3)

summary=nlargest(summarized_length,scored_sentences,key=scored_sentences.get) #highest scoring sentences 

print(len(sentences))
print(len(summary))

print(" \n".join(summary))

```
