```python
from transformers import pipeline

# Initialize the Hugging Face pipeline for summarization
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

# Function to ask user for input and summarize the text
def summarize_text():
    # Ask the user how many lines they want in the summary
    num_lines = int(input("How many lines of summary do you need? "))
    
    # Ask the user to paste their text
    print("Please paste your text for summarization (press Enter twice to finish):")
    text = ""
    while True:
        line = input()
        if line == "":
            break
        text += line + "\n"

    # Generate summary
    summary = summarizer(text, max_length=num_lines * 25, min_length=25, do_sample=False)

    # Output the summary
    print("\nSummary:")
    print(summary[0]['summary_text'])

# Call the function
summarize_text()
```
